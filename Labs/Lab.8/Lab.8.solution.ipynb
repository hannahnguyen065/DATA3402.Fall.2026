### Exercise 3: Training a Classifier

Let's repeat what we did by hand in the previous lab (Lab 7, Exercise 8) using scikit-learn. We'll use an LDA classifier, which we can instantiate as follows:

```python
import sklearn.discriminant_analysis as DA
Fisher=DA.LinearDiscriminantAnalysis()
```

As discussed in the lecture, to properly formulate our problem, we'll have to:

* Define the inputs (X) vs outputs (Y)
* Designate training vs testing samples (in order to get an unbiased assessment of the performance of machine learning algorithms)

For example, here we'll use 4M events for training and the remainder for testing.

```python
N_Train=4000000

Train_Sample=df[:N_Train]
Test_Sample=df[N_Train:]

X_Train=Train_Sample[VarNames[1:]]
y_Train=Train_Sample["signal"]

X_Test=Test_Sample[VarNames[1:]]
y_Test=Test_Sample["signal"]

Test_sig=Test_Sample[Test_Sample.signal==1]
Test_bkg=Test_Sample[Test_Sample.signal==0]
```

We can train the classifier as follows:

```python
Fisher.fit(X_Train,y_Train)
```

We can plot the output, comparing signal and background:

```python
plt.figure()
plt.hist(Fisher.decision_function(Test_sig[VarNames[1:]]),bins=100,histtype="step", color="blue", label="signal",stacked=True)
plt.hist(Fisher.decision_function(Test_bkg[VarNames[1:]]),bins=100,histtype="step", color="red", label="background",stacked=True)
plt.legend(loc='upper right')
plt.show()
```

#### Part a

Compare the ROC curves computed on the test and on the training samples, in a single plot. Do you see a bias? (Use `Fisher.decision_function` on each sample, and `roc_curve` as in Exercise 5.)

```python
from sklearn.metrics import roc_curve, auc

#Training sample
train_scores = Fisher.decision_function(X_Train)
fpr_train, tpr_train, _ = roc_curve(y_Train, train_scores)

#Testing sample
test_scores = Fisher.decision_function(X_Test)
fpr_test, tpr_test, _ = roc_curve(y_Test, test_scores)

#Calculate AUC
auc_train = auc(fpr_train, tpr_train)
auc_test = auc(fpr_test, tpr_test)

#Plot ROC curves
plt.figure()
plt.plot(fpr_train, tpr_train, label="Training AUC = %.3f" % auc_train)
plt.plot(fpr_test, tpr_test, label="Testing AUC = %.3f" % auc_test)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves for Training and Testing")
plt.legend()
plt.show()
```

The training and testing ROC curves are very similar with almost the same AUC values. This suggests that there is little to no noticeable overfitting since the model performs about the same on both datasets.

#### Part b

Train three Fisher discriminants, using as input (1) only the raw quantities (`RawNames`), (2) only the derived features (`FeatureNames`), and (3) all 18. Compare their ROC curves, computed on the test sample, in a single plot. Which inputs carry the separation?

```python
#Raw quantities
Fisher_raw = DA.LinearDiscriminantAnalysis()
Fisher_raw.fit(Train_Sample[RawNames], y_Train)

#Derived features
Fisher_features = DA.LinearDiscriminantAnalysis()
Fisher_features.fit(Train_Sample[FeatureNames], y_Train)

#All 18 features
Fisher_all = DA.LinearDiscriminantAnalysis()
Fisher_all.fit(X_Train, y_Train)


fpr_raw, tpr_raw, _ = roc_curve(y_Test, Fisher_raw.decision_function(Test_Sample[RawNames]))
fpr_features, tpr_features, _ = roc_curve(y_Test, Fisher_features.decision_function(Test_Sample[FeatureNames]))
fpr_all, tpr_all, _ = roc_curve(y_Test, Fisher_all.decision_function(X_Test))

plt.figure()
plt.plot(fpr_raw, tpr_raw, label="Raw AUC = %.3f" % auc(fpr_raw, tpr_raw))
plt.plot(fpr_features, tpr_features, label="Derived AUC = %.3f" % auc(fpr_features, tpr_features))
plt.plot(fpr_all, tpr_all, label="All AUC = %.3f" % auc(fpr_all, tpr_all))

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves")
plt.legend()
plt.show()
```

#### Part c

Use the best method from part b to compute the maximal significance $\sigma_S= \frac{N_S}{\sqrt{N_S+N_B}}$ for the four scenarios in Lab 7 (Exercise 5.3), choosing the cut on the classifier's output that maximizes it.

```python
#part A
from sklearn.linear_model import LogisticRegression
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.naive_bayes import GaussianNB

Logistic = LogisticRegression(max_iter=1000)
Fisher = LinearDiscriminantAnalysis()
NaiveBayes = GaussianNB()
```

```python
#part B
def compare_model(model, name):
    model.fit(X_Train, y_Train)

    if hasattr(model, "decision_function"):
        scores = model.decision_function(X_Test)
    else:
        scores = model.predict_proba(X_Test)[:, 1]

    fpr, tpr, _ = roc_curve(y_Test, scores)
    roc_auc = auc(fpr, tpr)

    plt.plot(fpr, tpr, label=name + " AUC = %.3f" % roc_auc)

    return roc_auc

plt.figure()

auc_logistic = compare_model(Logistic, "Logistic Regression")
auc_fisher = compare_model(Fisher, "LDA")
auc_nb = compare_model(NaiveBayes, "Naive Bayes")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Comparing Three Classifiers")
plt.legend()
plt.show()

print("Logistic Regression:", auc_logistic)
print("LDA:", auc_fisher)
print("Naive Bayes:", auc_nb)
```

```python
#part C

#Choosing the best classifier
results = {
    "Logistic Regression": auc_logistic,
    "LDA": auc_fisher,
    "Naive Bayes": auc_nb
}

best_name = max(results, key=results.get)
print("Best classifier:", best_name)

models = {
    "Logistic Regression": Logistic,
    "LDA": Fisher,
    "Naive Bayes": NaiveBayes
}

best_model = models[best_name]


#Finds the cutoff giving the highest significance
def find_significance(signal_total, background_total):
    if hasattr(best_model, "decision_function"):
        scores = best_model.decision_function(X_Test)
    else:
        scores = best_model.predict_proba(X_Test)[:, 1]

    fpr, tpr, thresholds = roc_curve(y_Test, scores)

    Ns = tpr * signal_total
    Nb = fpr * background_total

    significance = np.divide(
        Ns, np.sqrt(Ns + Nb),
        out=np.zeros_like(Ns),
        where=(Ns + Nb) > 0
    )

    best = np.argmax(significance)

    print("Best cutoff:", thresholds[best])
    print("Maximum significance:", significance[best])


#Four scenarios from Lab 7
scenarios = [
    (10, 100),
    (100, 1000),
    (1000, 10000),
    (10000, 100000)
]

for ns, nb in scenarios:
    print("NS:", ns, "NB:", nb)
    find_significance(ns, nb)


```
