*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
# Write your solution here
def check_game(board):
    n = len(board)

    #check rows
    for row in board:
        if row.count(1) == n:
            return 1
        if row.count(2) == n:
            return 2

    #check columns
    for col in range(n):
        column = []
        for row in range(n):
            column.append(board[row][col])

        if column.count(1) == n:
            return 1
        if column.count(2) == n:
            return 2

    #check first diagonal
    diagonal = []
    for i in range(n):
        diagonal.append(board[i][i])

    if diagonal.count(1) == n:
        return 1
    if diagonal.count(2) == n:
        return 2

    #check other diagonal
    diagonal = []
    for i in range(n):
        diagonal.append(board[i][n - 1 - i])

    if diagonal.count(1) == n:
        return 1
    if diagonal.count(2) == n:
        return 2

    #check for empty spaces
    for row in board:
        if 0 in row:
            return -1

    #if no winner and no empty spaces
    return 0
````

````python
# Write your solution here
def make_move(board, player, row, col):
    if board[row][col] == 0:
        board[row][col] = player
        return True
    else:
        return False
````

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,

- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
#solution
def play_game(n):
    board = []

    for i in range(n):
        board.append([0] * n)

    player = 1
    status = -1

    while status == -1:
        draw_game_labels(board)

        print("Player", player)
        player_move(board, player)

        status = check_game(board)

        if status == -1:
            if player == 1:
                player = 2
            else:
                player = 1

    draw_game_labels(board)

    if status == 0:
        print("The game is a draw.")
    elif status == 1:
        print("Player 1 (X) wins!")
    elif status == 2:
        print("Player 2 (O) wins!")
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
# Write your solution here
import random

def computer_move(board, player):
    n = len(board)

    if player == 1:
        opponent = 2
    else:
        opponent = 1

    #block opponent from winning
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = opponent

                if check_game(board) == opponent:
                    board[row][col] = player
                    return

                board[row][col] = 0

    #try to make a winning move
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = player

                if check_game(board) == player:
                    return

                board[row][col] = 0

    
    empty_spaces = []

    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                empty_spaces.append((row, col))

    if len(empty_spaces) > 0:
        move = random.choice(empty_spaces)
        board[move[0]][move[1]] = player
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
# Write your solution here
def search(board, player, computer, depth, max_depth):
    status = check_game(board)

    if status == computer:
        return 10

    if status != -1 and status != 0:
        return -10

    if status == 0:
        return 0

    if depth == max_depth:
        return 0

    n = len(board)

    if player == computer:
        best_score = -100

        for row in range(n):
            for col in range(n):
                if board[row][col] == 0:
                    board[row][col] = player

                    if player == 1:
                        next_player = 2
                    else:
                        next_player = 1

                    score = search(board, next_player, computer,
                                   depth + 1, max_depth)

                    board[row][col] = 0

                    if score > best_score:
                        best_score = score

        return best_score

    else:
        best_score = 100

        for row in range(n):
            for col in range(n):
                if board[row][col] == 0:
                    board[row][col] = player

                    if player == 1:
                        next_player = 2
                    else:
                        next_player = 1

                    score = search(board, next_player, computer,
                                   depth + 1, max_depth)

                    board[row][col] = 0

                    if score < best_score:
                        best_score = score

        return best_score
````

````python
def smart_computer_move(board, player, max_depth):
    n = len(board)

    best_score = -100
    best_move = None

    if player == 1:
        opponent = 2
    else:
        opponent = 1

    for row in range(n):
        for col in range(n):

            if board[row][col] == 0:
                board[row][col] = player

                score = search(board, opponent, player, 1, max_depth)

                board[row][col] = 0

                if score > best_score:
                    best_score = score
                    best_move = (row, col)

    if best_move != None:
        board[best_move[0]][best_move[1]] = player
````
